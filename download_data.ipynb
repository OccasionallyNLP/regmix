```python
def formatting(data, name):
    output = []
    for i in data:
        if name in ['gsm8k']:
            output.append(dict(question=i['question'], answer=i['answer']))
        elif name == 'mmlu':
            answer = i['answer']
            answer = i['choices'][answer]
            output.append(dict(question=i['question'], answer=answer)),
        elif name == 'hellaswag':
            label = int(i['label'])
            output.append(dict(question=i['ctx'], answer=i['endings'][label]))
        elif name == 'haerae':
            output.append(dict(question=i['query'],answer=i['answer']))
        elif name == 'kmmlu':
            answer2answer = {1:'A',2:'B',3:'C',4:'D'}
            output.append(dict(question=i['question'],answer=i[answer2answer[i['answer']]]))
        elif name == 'humaneval':
            output.append(dict(question=i['prompt'], answer=i['canonical_solution']))
        elif name == 'mbpp':
            output.append(dict(question=i['text'], answer=i['code']))
    output = Dataset.from_list(output) 
    return output
```
